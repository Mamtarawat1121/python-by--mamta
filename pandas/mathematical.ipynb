```python
import numpy as np
import pandas as pd
```

```python
data = [15,22,45,52,61,25,33,18]
```

```python
math_intervals = pd.cut(data, bins = 3)
```

```python
text_labels = pd.cut(data, bins = 3,labels = ["Low","Medium","High"])
```

```python
df = pd.DataFrame({
    "Original_value" : data,
    "Math_intervals" : math_intervals,
    "Text_Label" : text_labels
})
```
